# Loop

Loop 指令可以讓程式碼依我的條件不斷執行，直到條件滿足為止

```python
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

```python
for i in range(5):
    print('&'*(i+1))
```

```python
for i in range(10):
    print(f'1*{(i+1)}={1*(i+1)}')
```

# 九九乘法表

1\*1=1, 1\*2=2, 1\*3=3,...  
2\*1=2, 2\*2=4, 2\*3=6,...  
3\*1=3, 3\*2=6, 3\*3=9,...  

```python
i = 1
for j in range(1, 6):
    print(f'{i}*{j}={i*j}', end=',')
print()
i =2
for j in range(1, 6):
    print(f'{i}*{j}={i*j}', end=',')
print()
i = 3
for j in range(1, 6):
    print(f'{i}*{j}={i*j}', end=',')
print()
```

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(f'{i}*{j}={i*j}', end=',')
    print()
```

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(f'{i}*{j}={i*j}', end=',')
    print()
```

```python
n = 17
isPrime = True
for i  in range(2, n):
    if n % i == 0:
        print(f'{17}不是質數, 因為它可以被{i}除盡')
        isPrime = False
        break

if isPrime == True:
    print(f'{n}是質數')
```
