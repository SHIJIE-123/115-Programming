# 介紹

本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人

```python
import pyautogui as pyg
pyg.sleep(2)
basex = 430
basey = 385
gap = 40

```

```python
def get_position(r, c):
    x = basex + c * gap
    y = basey + r * gap
    return (x, y)
```

```python
def get_gem_colar(r, c, im):
    x, y = get_position(r, c)
    return im.getpixel((x, y))
```

```python
def can_move_left(r, c, im):
    if not (0 <= r < 8 and 1 <= c < 8):
        return False

    left = (r, c - 1)
    current = (r, c)

    def color_after_swap(row, col):
        if (row, col) == left:
            return get_gem_colar(*current, im)
        if (row, col) == current:
            return get_gem_colar(*left, im)
        return get_gem_colar(row, col, im)

    if get_gem_colar(*left, im) == get_gem_colar(*current, im):
        return False

    for row, col in (left, current):
        target_color = color_after_swap(row, col)
        for dr, dc in ((0, 1), (1, 0)):
            count = 1
            for sign in (-1, 1):
                nr, nc = row + sign * dr, col + sign * dc
                while 0 <= nr < 8 and 0 <= nc < 8 and color_after_swap(nr, nc) == target_color:
                    count += 1
                    nr += sign * dr
                    nc += sign * dc
            if count >= 3:
                return True

    return False
```

```python
def _can_move_to(r, c, dr, dc, im):
    nr, nc = r + dr, c + dc
    if not (0 <= r < 8 and 0 <= c < 8 and 0 <= nr < 8 and 0 <= nc < 8):
        return False

    first = (r, c)
    second = (nr, nc)

    def color_after_swap(row, col):
        if (row, col) == first:
            return get_gem_colar(*second, im)
        if (row, col) == second:
            return get_gem_colar(*first, im)
        return get_gem_colar(row, col, im)

    if get_gem_colar(*first, im) == get_gem_colar(*second, im):
        return False

    for row, col in (first, second):
        target_color = color_after_swap(row, col)
        for match_dr, match_dc in ((0, 1), (1, 0)):
            count = 1
            for sign in (-1, 1):
                check_r = row + sign * match_dr
                check_c = col + sign * match_dc
                while (0 <= check_r < 8 and 0 <= check_c < 8
                       and color_after_swap(check_r, check_c) == target_color):
                    count += 1
                    check_r += sign * match_dr
                    check_c += sign * match_dc
            if count >= 3:
                return True

    return False

def can_move_right(r, c, im):
    return _can_move_to(r, c, 0, 1, im)

def can_move_up(r, c, im):
    return _can_move_to(r, c, -1, 0, im)

def can_move_down(r, c, im):
    return _can_move_to(r, c, 1, 0, im)
```

```python
import time

start_time = time.time()
im = pyg.screenshot()
directions = (
    (can_move_left, 0, -1),
    (can_move_right, 0, 1),
    (can_move_up, -1, 0),
    (can_move_down, 1, 0),
)

while time.time() - start_time <= 60:
    moved = False
    for r in range(8):
        if time.time() - start_time > 60:
            break
        for c in range(8):
            if time.time() - start_time > 60:
                break
            for can_move, dr, dc in directions:
                if time.time() - start_time > 60:
                    break
                if can_move(r, c, im):
                    x1, y1 = get_position(r, c)
                    x2, y2 = get_position(r + dr, c + dc)
                    pyg.moveTo(x1, y1)
                    pyg.dragTo(x2, y2, duration=0.2)
                    pyg.sleep(0.2)
                    im = pyg.screenshot()
                    moved = True
                    break
            if moved:
                break
        if moved:
            break
    if not moved:
        pyg.sleep(0.05)
```
