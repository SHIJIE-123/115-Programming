# 介紹

本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人

```python
import pyautogui as pyg
pyg.sleep(2)
basex = 430
basey = 385
gap = 40

```

```python
for row in range(8):
    for col in range(8):
        x = basex + col * gap
        y = basey + row * gap
        color = pyg.pixel(x, y)
        print(f"第 {row + 1} 列、第 {col + 1} 欄 ({x}, {y})：{color}")
```
