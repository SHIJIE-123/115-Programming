```python
scores = [78, 29, 56, 80, 99]
print(scores)
```

## 索引值

我們可以透過索引值(index)來存取串列中的元素

注意:索引值由0開始

```python
print(scores[0])
print(scores[1])
```

## 串列內容的個數

len(a) 取得 a 這個串列中元素的個數

len 是 length (長度) 的意思

```python
print(len(scores))
print(scores[4])
print(scores[5-1])
print(scores[len(scores)-1])
print(scores[-1])
```
