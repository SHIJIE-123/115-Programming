```python
scores = [78, 29, 56, 80, 99]
print(scores)
```

## 索引值

我們可以透過索引值(index)來存取串列中的元素

注意:索引值由0開始

```python
print(scores[0])
print(scores[1])
```

## 串列內容的個數

len(a) 取得 a 這個串列中元素的個數

len 是 length (長度) 的意思

```python
print(len(scores))
print(scores[4])
print(scores[5-1])
print(scores[len(scores)-1])
print(scores[-1])
```

使用 `append()` 新增元素到串列的尾端

```python
print(scores)
scores.append(78)
print(scores)
```

```python
print(scores)
scores[2] = 70
print(scores)
```

```python
weights = [12]*6
print(weights)
```

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
for e in scores:
    print(e)
```

```python
for i, e in enumerate(scores):
    print(i, e)
```
